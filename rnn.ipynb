```python
sequence = [1, 2, 3, 1, 2, 3, 1, 2, 3]

# Step 4: build input -> target pairs
inputs = sequence[:-1]
targets = sequence[1:]
print("Inputs: ", inputs)
print("Targets:", targets)

# Vocabulary
vocab = sorted(set(sequence))
vocab_size = len(vocab)
num_to_idx = {num: idx for idx, num in enumerate(vocab)}
idx_to_num = {idx: num for idx, num in enumerate(vocab)}
print("Vocab:", vocab, "| size:", vocab_size)
print("num_to_idx:", num_to_idx)

# Step 5: one-hot encoding function
def one_hot(num, vocab_size, num_to_idx):
    vec = np.zeros(vocab_size)
    vec[num_to_idx[num]] = 1
    return vec

# Test it
print("One-hot for 2:", one_hot(2, vocab_size, num_to_idx))
```

```python
hidden_size = 4
vocab_size = 3  # already defined earlier, just confirming

np.random.seed(42)

Wxh = np.random.randn(hidden_size, vocab_size) * 0.01
Whh = np.random.randn(hidden_size, hidden_size) * 0.01
bh = np.zeros(hidden_size)

h_prev = np.zeros(hidden_size)  # h_0, the very first hidden state

print("Wxh shape:", Wxh.shape)
print("Whh shape:", Whh.shape)
print("bh shape:", bh.shape)
print("h_prev (h0):", h_prev)
```

```python
# Take the very first input in our sequence: the number 1
x_t = one_hot(1, vocab_size, num_to_idx)
print("x_t (one-hot for input '1'):", x_t)

# The equation: h_t = tanh(Wxh @ x_t + Whh @ h_prev + bh)
a_t = Wxh @ x_t + Whh @ h_prev + bh
print("a_t (before tanh):", a_t)

h_t = np.tanh(a_t)
print("h_t (after tanh):", h_t)
```

```python
output_size = vocab_size  # we predict one of {1,2,3}, so output size = 3

np.random.seed(42)
Why = np.random.randn(output_size, hidden_size) * 0.01
by = np.zeros(output_size)

print("Why shape:", Why.shape)
print("by shape:", by.shape)

# The equation: o_t = Why @ h_t + by
o_t = Why @ h_t + by
print("o_t (raw output scores):", o_t)
```

```python
def softmax(x):
    exp_x = np.exp(x)
    return exp_x / np.sum(exp_x)

probs = softmax(o_t)
print("Probabilities:", probs)
print("Sum of probabilities:", np.sum(probs))
```

```python
predicted_idx = np.argmax(probs)
predicted_num = idx_to_num[predicted_idx]

print("Predicted index:", predicted_idx)
print("Predicted number:", predicted_num)
```

```python
# Our input was 1, and the correct target (from our dataset) is 2
target_num = 2
target_idx = num_to_idx[target_num]

p_correct = probs[target_idx]
loss = -np.log(p_correct)

print("Target index:", target_idx)
print("Probability assigned to correct answer:", p_correct)
print("Loss:", loss)
```

```python
h_prev = np.zeros(hidden_size)  # reset hidden state to h0 before starting the sequence

hidden_states = []
outputs = []
probs_list = []
losses = []

for t in range(len(inputs)):
    x_t = one_hot(inputs[t], vocab_size, num_to_idx)
    target_idx = num_to_idx[targets[t]]

    # Forward pass, one timestep
    a_t = Wxh @ x_t + Whh @ h_prev + bh
    h_t = np.tanh(a_t)
    o_t = Why @ h_t + by
    p_t = softmax(o_t)
    loss_t = -np.log(p_t[target_idx])

    # Store everything (we'll need it for BPTT later)
    hidden_states.append(h_t)
    outputs.append(o_t)
    probs_list.append(p_t)
    losses.append(loss_t)

    # Carry hidden state forward to next timestep
    h_prev = h_t

    print(f"t={t} | input={inputs[t]} target={targets[t]} | predicted={idx_to_num[np.argmax(p_t)]} | loss={loss_t:.4f}")

total_loss = np.mean(losses)
print("\nTotal average loss:", total_loss)
```

```python
t = len(inputs) - 1  # last timestep
x_t = one_hot(inputs[t], vocab_size, num_to_idx)
target_idx = num_to_idx[targets[t]]
h_t = hidden_states[t]
p_t = probs_list[t]

# One-hot target vector
y_true = np.zeros(vocab_size)
y_true[target_idx] = 1

# Output error
delta_o = p_t - y_true
print("p_t:", p_t)
print("y_true:", y_true)
print("delta_o:", delta_o)

# Gradients for output layer
dWhy = np.outer(delta_o, h_t)
dby = delta_o

print("\ndWhy shape:", dWhy.shape, "(should match Why shape:", Why.shape, ")")
print("dby shape:", dby.shape)
```

```python
h_prev_t = hidden_states[t-1] if t > 0 else np.zeros(hidden_size)

# Step A + B: hidden error, accounting for tanh derivative
delta_h = (Why.T @ delta_o) * (1 - h_t**2)
print("delta_h:", delta_h)

# Step C: gradients for hidden-layer parameters
dWxh = np.outer(delta_h, x_t)
dWhh = np.outer(delta_h, h_prev_t)
dbh = delta_h

print("\ndWxh shape:", dWxh.shape, "(should match Wxh:", Wxh.shape, ")")
print("dWhh shape:", dWhh.shape, "(should match Whh:", Whh.shape, ")")
print("dbh shape:", dbh.shape)
```

```python
# Initialize gradient accumulators at zero
dWxh = np.zeros_like(Wxh)
dWhh = np.zeros_like(Whh)
dbh = np.zeros_like(bh)
dWhy = np.zeros_like(Why)
dby = np.zeros_like(by)

delta_h_next = np.zeros(hidden_size)  # no future error after the last timestep

for t in reversed(range(len(inputs))):
    x_t = one_hot(inputs[t], vocab_size, num_to_idx)
    target_idx = num_to_idx[targets[t]]
    h_t = hidden_states[t]
    h_prev_t = hidden_states[t-1] if t > 0 else np.zeros(hidden_size)
    p_t = probs_list[t]

    y_true = np.zeros(vocab_size)
    y_true[target_idx] = 1

    # Output error (Step 14)
    delta_o = p_t - y_true
    dWhy += np.outer(delta_o, h_t)
    dby += delta_o

    # Hidden error: own contribution + future contribution (Step 16 key equation)
    delta_h = (Why.T @ delta_o + Whh.T @ delta_h_next) * (1 - h_t**2)

    dWxh += np.outer(delta_h, x_t)
    dWhh += np.outer(delta_h, h_prev_t)
    dbh += delta_h

    delta_h_next = delta_h  # carry backward to the next (earlier) iteration

print("dWxh:\n", dWxh)
print("\ndWhh:\n", dWhh)
print("\ndbh:", dbh)
print("\ndWhy:\n", dWhy)
print("\ndby:", dby)
```

```python
predicted_idx = np.argmax(probs)
predicted_num = idx_to_num[predicted_idx]

print("Predicted index:", predicted_idx)
print("Predicted number:", predicted_num)
```

```python
print("Clipped dWxh max/min:", dWxh.max(), dWxh.min())
```

```python
learning_rate = 0.1

Wxh -= learning_rate * dWxh
Whh -= learning_rate * dWhh
bh  -= learning_rate * dbh
Why -= learning_rate * dWhy
by  -= learning_rate * dby

print("Updated Wxh:\n", Wxh)
```

```python
import numpy as np

# ============================================
# PHASE 2 — Dataset
# ============================================
sequence = [1, 2, 3, 1, 2, 3, 1, 2, 3]
inputs = sequence[:-1]
targets = sequence[1:]

vocab = sorted(set(sequence))
vocab_size = len(vocab)
num_to_idx = {num: idx for idx, num in enumerate(vocab)}
idx_to_num = {idx: num for idx, num in enumerate(vocab)}

def one_hot(num, vocab_size, num_to_idx):
    vec = np.zeros(vocab_size)
    vec[num_to_idx[num]] = 1
    return vec

# ============================================
# HELPER FUNCTIONS
# ============================================
def softmax(x):
    exp_x = np.exp(x)
    return exp_x / np.sum(exp_x)

def clip_gradients(grads, clip_value=5):
    return [np.clip(g, -clip_value, clip_value) for g in grads]

# ============================================
# PHASE 3 — Initialize Parameters
# ============================================
np.random.seed(42)
hidden_size = 4
learning_rate = 0.1
epochs = 2000

Wxh = np.random.randn(hidden_size, vocab_size) * 0.01
Whh = np.random.randn(hidden_size, hidden_size) * 0.01
Why = np.random.randn(vocab_size, hidden_size) * 0.01
bh = np.zeros(hidden_size)
by = np.zeros(vocab_size)

# ============================================
# PHASE 9 — Training Loop
# ============================================
loss_history = []

for epoch in range(epochs):
    h_prev = np.zeros(hidden_size)
    hidden_states = []
    probs_list = []
    losses = []

    # ---- FORWARD PASS ----
    for t in range(len(inputs)):
        x_t = one_hot(inputs[t], vocab_size, num_to_idx)
        target_idx = num_to_idx[targets[t]]

        a_t = Wxh @ x_t + Whh @ h_prev + bh
        h_t = np.tanh(a_t)
        o_t = Why @ h_t + by
        p_t = softmax(o_t)
        loss_t = -np.log(p_t[target_idx])

        hidden_states.append(h_t)
        probs_list.append(p_t)
        losses.append(loss_t)
        h_prev = h_t

    total_loss = np.mean(losses)
    loss_history.append(total_loss)

    # ---- BACKWARD PASS (BPTT) ----
    dWxh = np.zeros_like(Wxh)
    dWhh = np.zeros_like(Whh)
    dbh = np.zeros_like(bh)
    dWhy = np.zeros_like(Why)
    dby = np.zeros_like(by)
    delta_h_next = np.zeros(hidden_size)

    for t in reversed(range(len(inputs))):
        x_t = one_hot(inputs[t], vocab_size, num_to_idx)
        target_idx = num_to_idx[targets[t]]
        h_t = hidden_states[t]
        h_prev_t = hidden_states[t-1] if t > 0 else np.zeros(hidden_size)
        p_t = probs_list[t]

        y_true = np.zeros(vocab_size)
        y_true[target_idx] = 1

        delta_o = p_t - y_true
        dWhy += np.outer(delta_o, h_t)
        dby += delta_o

        delta_h = (Why.T @ delta_o + Whh.T @ delta_h_next) * (1 - h_t**2)
        dWxh += np.outer(delta_h, x_t)
        dWhh += np.outer(delta_h, h_prev_t)
        dbh += delta_h

        delta_h_next = delta_h

    # ---- CLIP ----
    dWxh, dWhh, dbh, dWhy, dby = clip_gradients([dWxh, dWhh, dbh, dWhy, dby])

    # ---- UPDATE ----
    Wxh -= learning_rate * dWxh
    Whh -= learning_rate * dWhh
    bh  -= learning_rate * dbh
    Why -= learning_rate * dWhy
    by  -= learning_rate * dby

    if epoch % 200 == 0:
        print(f"Epoch {epoch:4d} | Loss: {total_loss:.4f}")

print(f"\nFinal Loss: {loss_history[-1]:.4f}")

# ============================================
# PHASE 10 — Test the Trained Model
# ============================================
print("\n--- Testing the trained RNN ---")
h_prev = np.zeros(hidden_size)

for test_input in [1, 2, 3]:
    x_t = one_hot(test_input, vocab_size, num_to_idx)
    a_t = Wxh @ x_t + Whh @ h_prev + bh
    h_t = np.tanh(a_t)
    o_t = Why @ h_t + by
    p_t = softmax(o_t)
    predicted_num = idx_to_num[np.argmax(p_t)]

    print(f"Input: {test_input} -> Predicted: {predicted_num} | Probabilities: {p_t}")
    h_prev = h_t
```

```python
print("\n==============================")
print("PHASE 11 — Sequence Generation")
print("==============================")

start_num = 1          # Starting number
steps = 10             # How many numbers to generate

generated = [start_num]

h_prev = np.zeros(hidden_size)
current_num = start_num

for _ in range(steps):

    # Convert current number to one-hot
    x_t = one_hot(current_num, vocab_size, num_to_idx)

    # Forward pass
    a_t = Wxh @ x_t + Whh @ h_prev + bh
    h_t = np.tanh(a_t)

    o_t = Why @ h_t + by
    p_t = softmax(o_t)

    # Predict next number
    predicted_idx = np.argmax(p_t)
    predicted_num = idx_to_num[predicted_idx]

    generated.append(predicted_num)

    # Feed prediction back as next input
    current_num = predicted_num
    h_prev = h_t

print("Generated sequence:")
print(generated)
```

```python
def predict(num):

    if num not in [1, 2, 3]:
        print("Invalid input!")
        return

    h_prev = np.zeros(hidden_size)

    x_t = one_hot(num, vocab_size, num_to_idx)

    a_t = Wxh @ x_t + Whh @ h_prev + bh
    h_t = np.tanh(a_t)

    o_t = Why @ h_t + by
    p_t = softmax(o_t)

    predicted_idx = np.argmax(p_t)
    predicted_num = idx_to_num[predicted_idx]

    print(f"Input: {num}")
    print(f"Prediction: {predicted_num}")
    print(f"Probabilities: {p_t}")
```
